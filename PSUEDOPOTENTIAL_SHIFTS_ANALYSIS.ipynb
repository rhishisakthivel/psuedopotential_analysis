```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from ipywidgets import interact
```

```python
q = 1.602176634e-19

Ey_90 = 0.15         

V = np.linspace(0, -2000, 100)

Ey = Ey_90 * (V / 90)

Ey_DC = Ey

plt.figure(figsize=(12,8))
plt.plot(V, Ey_DC, linewidth=2)

plt.xlabel("Applied Tip Voltage (V)")
plt.ylabel(r"$E_y^{(\mathrm{DC})}$ (V/mm)")
plt.title(r"Linear Effective Potential Coefficient from DC Tip")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
q = 1.602176634e-19
m = 9.1093837015e-31
Omega = 1.6e9 * (2 * np.pi)

U2_90 = 202.77      
Ey2_90 = 3.47        

V = np.linspace(0, 100, 20)

U1_array = U2_90 * (V / 90) * 1e6
Ey_array = Ey2_90 * (V / 90) * 1e3

PEy_array = (q / (2 * m * Omega**2)) * U1_array * Ey_array / 1000

def plot(v):
    U1 = U2_90 * (v / 90) * 1e6
    Ey = Ey2_90 * (v / 90) * 1e3

    PEy = (q / (2 * m * Omega**2)) * U1 * Ey / 1000

    plt.figure(figsize=(12,8))
    plt.plot(V, PEy_array, linewidth=2)
    plt.scatter(v, PEy, color='red', s=80)

    plt.xlabel("Applied RF Voltage (V)")
    plt.ylabel(r"$PE_y$ Coefficient (V/mm)")
    plt.title(r"RF Pseudopotential $PE_y$ Coefficient for Bare Trap")
    plt.grid(True)

    print(f"Voltage = {v} V")
    print(f"PEy = {PEy} V/mm")

    plt.show()

interact(plot, v=(0, 100, 1))
```

```python
q = 1.602176634e-19
m = 9.1093837015e-31
Omega = 1.6e9 * (2 * np.pi)

V = np.linspace(0, 100, 20)

U1_90_1 = 200.92      
Ey_90_1 = 2.88        

U1_90_2 = 202.77       
Ey_90_2 = 3.47         

U1_array1 = U1_90_1 * (V / 90) * 1e6
Ey_array1 = Ey_90_1 * (V / 90) * 1e3
PEy_array1 = (q / (2 * m * Omega**2)) * U1_array1 * Ey_array1 / 1000

U1_array2 = U1_90_2 * (V / 90) * 1e6
Ey_array2 = Ey_90_2 * (V / 90) * 1e3
PEy_array2 = (q / (2 * m * Omega**2)) * U1_array2 * Ey_array2 / 1000

PEy_diff_array = PEy_array1 - PEy_array2

def plot(v):
    U1_1 = U1_90_1 * (v / 90) * 1e6
    Ey_1 = Ey_90_1 * (v / 90) * 1e3
    PEy_1 = (q / (2 * m * Omega**2)) * U1_1 * Ey_1 / 1000

    U1_2 = U1_90_2 * (v / 90) * 1e6
    Ey_2 = Ey_90_2 * (v / 90) * 1e3
    PEy_2 = (q / (2 * m * Omega**2)) * U1_2 * Ey_2 / 1000
    
    PEy_diff = PEy_1 - PEy_2

    plt.figure(figsize=(12, 8))
    plt.plot(V, PEy_diff_array, linewidth=2, color='purple', label=r'$PEy_{full} - PEy_{bare}$')
    plt.scatter(v, PEy_diff, color='red', s=80, zorder=5, label=f'Current $V$ = {v} V')

    plt.xlabel("Applied RF Voltage (V)")
    plt.ylabel(r"$\Delta PE_y$ Coefficient (V/mm)")
    plt.title(r"Difference in RF Pseudopotential $PE_y$ Coefficient ($PEy_{full} - PEy_{bare}$)")
    plt.grid(True)
    plt.legend()

    print(f"Voltage = {v} V")
    print(f"PEy_full - PEy_bare = {PEy_diff} V/mm")

    plt.show()

interact(plot, v=(0, 100, 1))
```

```python
q = 1.602176634e-19
m = 9.1093837015e-31
Omega = 1.6e9 * (2 * np.pi)

U1_90 = 200.92
RF_Ey_90 = 2.88

Tip_Ey_90 = 0.15

tip_V = np.linspace(0, -2000, 201)
rf_V = np.linspace(0, 100, 201)

TIP, RF = np.meshgrid(tip_V, rf_V)

Tip_Ey = Tip_Ey_90 * (TIP / 90)

U1 = U1_90 * (RF / 90) * 1e6
RF_Ey = RF_Ey_90 * (RF / 90) * 1e3

PEy = (q / (2 * m * Omega**2)) * U1 * RF_Ey / 1000

Difference = PEy - Tip_Ey

plt.figure(figsize=(10,8))

plt.imshow(
    Difference,
    origin='lower',
    extent=[0, -2000, 0, 100],
    aspect='auto',
    cmap='bwr'
)

plt.colorbar(label="RF PEy − DC Ey")

plt.xlabel("Tip Voltage (V)")
plt.ylabel("RF Voltage (V)")
plt.title("RF PEy − DC Ey")

plt.tight_layout()
plt.show()
```

```python
q= 1.602176634e-19
m = 9.1093837015e-31
Omega = 1.6e9 * (2 * np.pi)

U1_90 = 200.92
RF_Ey_90 = 2.88
Tip_Ey_90 = 0.15

V_default = 90

tip_V = np.linspace(0, -2000, 201)
rf_V = np.linspace(0, 100, 201)

TIP, RF = np.meshgrid(tip_V, rf_V)

Tip_Ey = Tip_Ey_90 * (TIP / 90)

U1 = U1_90 * (RF / 90) * 1e6
RF_Ey = RF_Ey_90 * (RF / 90) * 1e3

PEy = (q / (2 * m * Omega**2)) * U1 * RF_Ey / 1000

Difference = PEy - Tip_Ey

kappa = rf_V / V_default
C = (q**2 / (m * Omega**2)) * (U1_90 * 1e6 * kappa)**2
omega = np.sqrt(2 * C / m)
f_MHz = omega / (2 * np.pi) / 1e6

plt.figure(figsize=(10, 8))

plt.imshow(
    Difference,
    origin='lower',
    extent=[0, -2000, 0, len(rf_V) - 1],
    aspect='auto',
    cmap='bwr',
    vmin=np.min(Difference),
    vmax=np.max(Difference)
)

tick_idx = np.linspace(0, len(rf_V) - 1, 6).astype(int)

plt.yticks(
    tick_idx,
    [f"{f_MHz[i]:.2f}" for i in tick_idx]
)

plt.colorbar(label=r"$PE_y - E_y^{DC}$ (V/mm)")

plt.xlabel("Tip Voltage (V)")
plt.ylabel("Secular Frequency (MHz)")
plt.title(r"$PE_y - E_y^{DC}$")

plt.tight_layout()
plt.show()
```
